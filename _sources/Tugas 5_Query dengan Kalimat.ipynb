```python
import math
import numpy as np
import string
import json
import pandas as pd
import re
import time
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
df = pd.read_csv("hasil_stemming_stopwords.csv")
df.head(5)
```

```python
# Transformasi TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(df['final_text'])
```

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

# Mengonversi TF-IDF matrix ke dalam DataFrame
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())

# Menampilkan hasil TF-IDF dalam bentuk DataFrame
print(tfidf_df)
```

```python
from sklearn.decomposition import TruncatedSVD

# Transformasi SVD
svd = TruncatedSVD(n_components=100)  # Ubah n_components sesuai kebutuhan
tfidf_svd = svd.fit_transform(tfidf_matrix)
```

```python
# Mengonversi hasil SVD ke dalam DataFrame
svd_df = pd.DataFrame(tfidf_svd)

# Menampilkan hasil
print(svd_df)
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.metrics.pairwise import cosine_similarity

# Standardisasi hasil SVD
scaler = StandardScaler()  # Impor yang diperlukan
ss_svd = scaler.fit_transform(tfidf_svd)
```

```python
# Mengonversi hasil SVD ke dalam DataFrame
ss_df = pd.DataFrame(ss_svd)

# Menampilkan hasil
print(ss_df)
```

```python
# Fungsi pencarian dokumen
def search_documents(query):
    # Transformasi query ke bentuk TF-IDF dan SVD
    query_tfidf = tfidf_vectorizer.transform([query])
    query_svd = svd.transform(query_tfidf)
    query_svd = scaler.transform(query_svd)  # Standardisasi query

    # Menghitung cosine similarity antara query dan setiap dokumen
    similarities = cosine_similarity(query_svd, tfidf_svd)
    df['Similarity'] = similarities[0]

    # Menampilkan 5 dokumen paling mirip
    return df.sort_values(by='Similarity', ascending=False).head(5)[['final_text', 'Similarity']]
```

```python
import pandas as pd

def highlight_text(text, query):
    # Menggunakan ANSI escape code untuk menyorot teks dengan warna biru
    return text.replace(query, f"\033[1;34m{query}\033[0m")  # Warna biru

# Fungsi untuk mencetak hasil pencarian
def print_search_results(results, query):
    if results.empty:  # Periksa apakah DataFrame kosong
        print("Tidak ada hasil yang ditemukan.")
    else:
        print("Hasil Pencarian:")
        for idx, result in results.iterrows():  # Iterasi melalui setiap baris DataFrame
            highlighted_text = highlight_text(result['final_text'], query)  # Mengganti 'final_text' dengan kolom yang benar
            print(f"{idx + 1}. {highlighted_text}")

# Input query dari pengguna
query = input("Masukkan kalimat pencarian: ")  # Meminta pengguna untuk memasukkan query
search_results = search_documents(query)  # Mencari dokumen berdasarkan query

# Mencetak hasil pencarian
print_search_results(search_results, query)
```
